```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# Load the shapefile
shapefile_path = '/Users/spencerzhang/Documents/PhD/Classes/F24 CS229/229 project/NaturalGas_InterIntrastate_Pipelines_US_EIA_6413543301445314001/NaturalGas_InterIntrastate_Pipelines_US_EIA.shp'
gdf = gpd.read_file(shapefile_path)
```

```python
# Plot the data
gdf.plot(figsize=(10, 10), edgecolor='black')

# Set plot title and labels
plt.title('Natural Gas Inter and Intrastate Pipelines in the US')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

# Show the plot
plt.show()
```
